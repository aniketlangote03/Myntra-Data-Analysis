# Data Quality Assessment

## Objective

Evaluate the quality of the Myntra dataset by checking for:

- Missing values
- Duplicate records
- Invalid values
- Incorrect data types
- Outliers
- Data consistency

The goal is to ensure the dataset is reliable before performing analysis.

## Import Required Libraries

Import the libraries required for data quality assessment.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

Load the Myntra dataset into a Pandas DataFrame.

```python
df = pd.read_csv("../data/myntra_dataset_ByScraping.csv")
```

## Verify Dataset

Display the first five rows to ensure that the dataset has been loaded correctly.

```python
df.head()
```

## Data Quality Check 1 — Missing Values

### Business Question

Are there any missing values in the dataset?

Missing values can affect calculations such as averages, correlations, and visualizations. Identifying them is the first step in any data quality assessment.

```python
df.isnull().sum()
```

### Observation

No missing values were found in any column of the dataset.

This indicates that all 52,120 product records contain complete information for brand name, product description, price, MRP, discount percentage, ratings, and number of ratings.

No data imputation or removal of missing records is required.

## Data Quality Check 2 — Missing Value Percentage

Knowing the percentage of missing values helps determine whether data should be removed or imputed.

```python
(df.isnull().sum()/len(df))*100
```

### Observation

The percentage of missing values is 0% across all columns.

Since every feature contains complete data, the dataset is suitable for further analysis without requiring missing value treatment.

## Data Quality Check 3 — Duplicate Rows

### Business Question

Are there duplicate records?

Duplicate records can inflate product counts and produce misleading business insights.

```python
df.duplicated().sum()
```

### Observation

A total of 17,047 duplicate rows were identified in the dataset.

This is a significant number of duplicates and requires further investigation before removal.

Since this is an e-commerce dataset, duplicate rows may represent products listed multiple times due to different sizes, colors, or repeated scraping. Therefore, these records should not be removed without verifying whether they are true duplicates.

## Data Quality Check 4 — Duplicate Product Descriptions

Some product descriptions may repeat because products are available in different sizes or colors.

We will check how many descriptions are repeated before deciding whether any action is needed.

```python
df["pants_description"].duplicated().sum()
```

### Observation

A total of **49,440 duplicate product descriptions** were found.

This result is expected in an e-commerce dataset because the same product description can be associated with multiple product variants such as different sizes, colors, or repeated listings.

Since the dataset does not contain attributes like size or color, duplicate product descriptions should **not** be considered data quality issues.

Therefore, no records will be removed based solely on duplicate product descriptions.

## Data Quality Check 5 — Verify Data Types

Correct data types are important because numerical analysis can only be performed on numeric columns.

```python
df.dtypes
```

### Observation

The dataset contains both categorical and numerical variables.

Categorical columns:
- brand_name
- pants_description

Numerical columns:
- price
- MRP
- discount_percent
- ratings
- number_of_ratings

The data types are appropriate for analysis and do not require conversion.

## Data Quality Check 6 — Validate Price

Business Rule:

Product price cannot be zero or negative.

```python
(df["price"] <= 0).sum()
```

### Observation

Invalid prices found: 0

## Data Quality Check 7 — Validate MRP

Business Rule:

MRP should always be greater than zero.

```python
(df["MRP"] <= 0).sum()
```

### Observation

No products were found with an MRP less than or equal to zero.

All products have valid original prices.

## Data Quality Check 8 — Price vs MRP

Business Rule

Selling Price should never exceed the MRP.

```python
(df["price"] > df["MRP"]).sum()
```

### Observation

No products were found where the selling price exceeds the MRP.

This confirms that the pricing information follows the expected business rule and appears to be internally consistent.

No corrective action is required.

## Data Quality Check 9 — Ratings

Business Rule

Ratings must be between 0 and 5.

```python
((df["ratings"] < 0) | (df["ratings"] > 5)).sum()
```

### Observation

All product ratings fall within the valid range of 0 to 5.

No invalid customer rating values were detected.

## Data Quality Check 10 — Discount Validation

Business Rule

Discount percentage should be between 0 and 1 because the dataset stores discount as a decimal value.

```python
((df["discount_percent"] < 0) | (df["discount_percent"] > 1)).sum()
```

### Observation

A total of **4,117 records** have discount values outside the expected decimal range of 0–1 (i.e., values greater than 1).

These values (ranging from 1.1 to 64.0) reflect an **inconsistent scraper format** — not standard percentage decimals (0.45) or whole percentages that can be reliably converted (e.g., dividing by 100 does not match price-derived discounts).

This issue will be addressed during the data cleaning phase.

### Observation

No products were found with a negative number of customer ratings.

All values in the `number_of_ratings` column satisfy the expected business rule and are valid for further analysis.

No corrective action is required.
